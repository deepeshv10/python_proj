Python decorators allow us to add additional capabilities to existing functions, without changing the original functions.
In this practice, I will work on python decorators. Starting with very basic implementation s below program :

```python

def decorator_method(print_name):
    def wrapper_function():
        print("Here is the name")
        return print_name()
    return wrapper_function

@decorator_method
def print_name():
    print("My name is Deepesh")

print_name()
```

If we want to use same decorator across multiple functions, say print_name() and print_full_details(name, age)
So we need to pass *args & **kwargs to the wrapper function arguments and also to the returned original function arguments.

```python
def decorator_method(original_function):
    def wrapper_function(*args, **kwargs):
        print("Here is the name")
        return original_function(*args, **kwargs)
    return wrapper_function

@decorator_method
def print_name():
    print("My name is Deepesh")

@decorator_method
def print_full_details(name, age):
    print(f"My name is {name} and Age is {age}")

print_name()
print_full_details("Deepesh", 30)
```

We can also implement decorator class, and apply this decorator functionalities to our function.
see the example below :

```python
class decorator_class(object):
    def __init__(self, original_function):
        self.original_function = original_function
    
    def __call__(self, *args, **kwargs):
        print("Here is the name")
        return self.original_function(*args, **kwargs)
    

@decorator_class
def print_name():
    print("My name is Deepesh")

@decorator_class
def print_full_details(name, age):
    print(f"My name is {name} and Age is {age}")

print_name()
print_full_details("Deepesh", 30)
```
