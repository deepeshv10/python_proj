In this practice, I am covering different methods that can be used in ThreadPoolExecutor

```python
# basic usage of TPE with submit method
# 
# 

from concurrent.futures import ThreadPoolExecutor
import time

def some_func(sec):
    print(f"Starting. ")
    time.sleep(sec)
    return f"Completed. "


def main():
    with ThreadPoolExecutor() as executor:
        f1 = executor.submit(some_func, 1)
        f2 = executor.submit(some_func, 1)
    
    print(f1.result())
    print(f2.result())


if __name__ == "__main__":
    main()
    

```

If we have multiple threads to be submitted, we can loop through each. To collect the results we will use 'as_completed' method. Below example shows this.

```python
# basic usage of TPE with submit method
# 
# 

from concurrent.futures import ThreadPoolExecutor, as_completed
import time

def some_func(sec):
    print(f"Starting. {sec}")
    time.sleep(sec)
    return f"Completed. {sec}"


def main():
    with ThreadPoolExecutor() as executor:
        results = [executor.submit(some_func, i) for i in range(5)]
    
        for result in as_completed(results):
            print(result.result())


if __name__ == "__main__":
    main()
    

```

Below prog. is same as above, but with reversed sleep times, to know the order of results

```python



from concurrent.futures import ThreadPoolExecutor, as_completed
import time

def some_func(sec):
    print(f"Starting. {sec}")
    time.sleep(sec)
    return f"Completed. {sec}"


def main():
    secs = [5,4,3,2,1]
    with ThreadPoolExecutor() as executor:
        results = [executor.submit(some_func, i) for i in secs]
    
        for result in as_completed(results):
            print(result.result())


if __name__ == "__main__":
    main()
    

```

Now we'll use the map method, to map the function with list of arguments

Notice the output of this, here even though the thread with sleep < 5secs would have completed first, but the map function will produce the results in the order of their start

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import time

def some_func(sec):
    print(f"Starting. {sec}")
    time.sleep(sec)
    return f"Completed. {sec}"


def main():
    secs = [5,4,3,2,1]
    with ThreadPoolExecutor() as executor:
        results = executor.map(some_func, secs)
    
        for result in results:
            print(result)


if __name__ == "__main__":
    main()
    

```

In below example I have same as above program but with use of max_workers (to limit the number of paralle threads)

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import time

def some_func(sec):
    print(f"Starting. {sec}")
    time.sleep(sec)
    return f"Completed. {sec}"


def main():
    secs = [5,4,3,2,1]
    with ThreadPoolExecutor(max_workers=2) as executor:     # max_workers=2, to limit max 2 threads at a time
        results = executor.map(some_func, secs)
    
        for result in results:
            print(result)


if __name__ == "__main__":
    main()
    
```
