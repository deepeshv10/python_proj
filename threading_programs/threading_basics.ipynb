Basic Threading example

```python
import threading
import time

def some_func(k):
    for i in range(2):
        print('Hello from thread')
        time.sleep(k)
        
def main():
    # single thread creation and invocation
    thread = threading.Thread(target=some_func(1))
    thread.start()
    thread.join()
    print("Main with single thread finished")
    
    # multiple threads creation and invocation, waiting untill all complete to continue with main
    threads = []
    for i  in  range(2):
        thread = threading.Thread(target = some_func(i))
        threads.append(thread)
        thread.start()
    
    for thread in threads:
        thread.join()
    
    print("Main with multi thread finished")
        
if __name__ == "__main__":
    main()
```

Example to show transferring data between two threads using a queue

publisher take 1 sec to put in a queue, but subscriber takes 3 sec to fetch from queue - this example demonstrates it as well

```python
import threading
import time
import queue

def publisher(queue):
    # puts an item in the queue, and sleeps for 1 sec
    for i in range(4):
        item = f"item-{i}"
        queue.put(item)
        print(f"{item} was put in queue by Publisher")
        time.sleep(1)
    queue.put(None)
    

def subscriber(queue):
    # gets an item from queue, and sleeps for 2 sec
    while True:
        item = queue.get()
        if item is None:
            break
        time.sleep(3)
        print(f"{item} was get from queue by Subscriber")

def main():
    # initialize threads for each publisher & subs
    # start them, join them
    q = queue.Queue()
    pubthread = threading.Thread(target=publisher, args=(q,))
    subthread = threading.Thread(target=subscriber, args=(q,))

    pubthread.start()
    subthread.start()
    
    pubthread.join()
    subthread.join()



if __name__ == "__main__":
    main()

```

Example usage of : ThreadPoolExecutor
Purpose is to get rid of manually managing the treads, threadpoolexecutor takes care of managing the threads for us

```python
from concurrent.futures import ThreadPoolExecutor
import time

def some_func(n):
    print(f'Task {n} will be handled by {threading.current_thread().name}')
    print(f'Task {n} started')
    time.sleep(n)
    print(f'Task {n} ended')
    return n*n

        
def main():
    # single thread creation and invocation
    with ThreadPoolExecutor(max_workers=3) as executor:
        results = executor.map(some_func, [4,2,3,1])
    
    print("Results :", list(results))
    print("Main with TPE finished")
        
if __name__ == "__main__":
    main()
```
