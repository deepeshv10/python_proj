## 1. Series & DataFrame — Creating from Scratch

A **Series** is a 1D labeled array. A **DataFrame** is a 2D labeled table (like a spreadsheet or SQL table). These are the two core data structures in pandas.

```python
# Series from a list
s1 = pd.Series([10, 20, 30, 40, 50])
print("Series from list:")
print(s1)
print(f"\nType: {type(s1)}")
print(f"dtype: {s1.dtype}")
print(f"Index: {s1.index.tolist()}")
```

```python
# Series with custom index
s2 = pd.Series(
    [85, 92, 78, 95],
    index=['Alice', 'Bob', 'Charlie', 'Diana'],
    name='test_score'
)
print("Series with custom index:")
print(s2)
print(f"\nAccess by label: s2['Bob'] = {s2['Bob']}")
print(f"Access by position: s2.iloc[1] = {s2.iloc[1]}")
```

```python
# Series from a dictionary
populations = pd.Series({
    'Tokyo': 13960000,
    'Delhi': 11030000,
    'Shanghai': 24870000,
    'Mumbai': 12440000
}, name='population')
print("Series from dict:")
print(populations)
```

## 2. Reading & Writing Data

pandas can read/write many formats: CSV, Excel, JSON, SQL, Parquet, and more. We'll demonstrate using in-memory data saved to temporary files.

```python
import tempfile, os

# Create a sample DataFrame
df_io = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'score': [85.5, 92.0, 78.3, 95.1, 88.7],
    'date': pd.date_range('2024-01-01', periods=5)
})

tmp_dir = tempfile.gettempdir()
```

```python
# Write to CSV and read back
csv_path = os.path.join(tmp_dir, 'sample.csv')
df_io.to_csv(csv_path, index=False)

df_from_csv = pd.read_csv(csv_path)
print("Read from CSV:")
print(df_from_csv)
print(f"\ndtypes:\n{df_from_csv.dtypes}")
```

```python
# read_csv with useful parameters
df_parsed = pd.read_csv(
    csv_path,
    parse_dates=['date'],     # auto-parse date column
    dtype={'id': 'int32'},    # specify dtypes to save memory
    usecols=['id', 'name', 'score']  # read only specific columns
)
print("Parsed CSV with options:")
print(df_parsed)
print(f"\ndtypes:\n{df_parsed.dtypes}")
```

```python
# Write to JSON and read back
json_path = os.path.join(tmp_dir, 'sample.json')
df_io.to_json(json_path, orient='records', indent=2, date_format='iso')

df_from_json = pd.read_json(json_path)
print("Read from JSON:")
print(df_from_json)
```

```python
# Clipboard — read from / write to system clipboard (interactive use)
# df.to_clipboard(index=False)   # copy DataFrame to clipboard
# df = pd.read_clipboard()       # paste from clipboard into DataFrame

# Clean up temp files
for f in [csv_path, json_path]:
    if os.path.exists(f):
        os.remove(f)

print("Common read functions:")
print("  pd.read_csv()     — CSV files")
print("  pd.read_excel()   — Excel files (.xlsx)")
print("  pd.read_json()    — JSON files")
print("  pd.read_sql()     — SQL databases")
print("  pd.read_parquet() — Parquet files (fast, columnar)")
print("  pd.read_html()    — HTML tables from web pages")
```

## 4. Selecting & Indexing — loc, iloc, boolean indexing

- **`[]`** — select columns by name, or rows by boolean mask
- **`loc`** — select by **label** (row/column names)
- **`iloc`** — select by **integer position** (0-based)
- **Boolean indexing** — filter rows based on conditions

```python
df_sel = df_inspect.copy()
df_sel
```

```python
# Select a single column (returns a Series)
print("Single column (Series):")
print(df_sel['name'])

# Select multiple columns (returns a DataFrame)
print("\nMultiple columns (DataFrame):")
print(df_sel[['name', 'salary']])
```

```python
# loc — label-based selection: df.loc[row_labels, column_labels]
print("Row 0, all columns:")
print(df_sel.loc[0])

print("\nRows 1-3, specific columns:")
print(df_sel.loc[1:3, ['name', 'age', 'salary']])
```

```python
# iloc — integer position-based selection
print("First 3 rows, first 2 columns:")
print(df_sel.iloc[:3, :2])

print("\nLast row:")
print(df_sel.iloc[-1])

print("\nEvery other row:")
print(df_sel.iloc[::2, :3])
```

```python
# Boolean indexing — filter rows with conditions
print("Salary > 80000:")
print(df_sel[df_sel['salary'] > 80000])

print("\nEngineering department with salary > 80000:")
mask = (df_sel['department'] == 'Engineering') & (df_sel['salary'] > 80000)
print(df_sel[mask])
```

```python
# isin() — filter by a list of values
print("Employees in Sales or HR:")
print(df_sel[df_sel['department'].isin(['Sales', 'HR'])])

# between() — range filter
print("\nSalary between 70000 and 90000:")
print(df_sel[df_sel['salary'].between(70000, 90000)])
```

```python
# Combining loc with boolean indexing
print("Names and salaries of managers:")
print(df_sel.loc[df_sel['is_manager'], ['name', 'salary']])
```

## 7. Handling Missing Data

Missing data is represented as `NaN` (Not a Number) or `NaT` (Not a Time). pandas provides tools to detect, remove, and fill missing values.

```python
df_missing = pd.DataFrame({
    'name': ['Alice', 'Bob', None, 'Diana', 'Eve'],
    'age': [25, np.nan, 35, 28, np.nan],
    'score': [88.5, 92.0, np.nan, np.nan, 76.0],
    'city': ['NYC', 'London', 'Paris', None, 'Sydney']
})
df_missing
```

```python
# Detect missing values
print("Is each cell null?")
print(df_missing.isna())

print("\nNull count per column:")
print(df_missing.isna().sum())

print(f"\nTotal nulls in DataFrame: {df_missing.isna().sum().sum()}")
print(f"Any nulls at all? {df_missing.isna().any().any()}")
```

```python
# dropna — remove rows/columns with missing values
print("Drop rows with ANY null:")
print(df_missing.dropna())

print("\nDrop rows where ALL values are null:")
print(df_missing.dropna(how='all'))

print("\nDrop rows with null in specific columns:")
print(df_missing.dropna(subset=['name', 'age']))
```

```python
# fillna — replace missing values
print("Fill with a constant:")
print(df_missing.fillna({'age': 0, 'score': df_missing['score'].mean(), 'city': 'Unknown'}))

print("\nForward fill (propagate previous value):")
print(df_missing.ffill())

print("\nBackward fill:")
print(df_missing.bfill())
```

```python
# interpolate — estimate missing values (useful for numeric/time series)
s_gaps = pd.Series([1, np.nan, np.nan, 4, 5, np.nan, 7])
print("Original:", s_gaps.values)
print("Interpolated:", s_gaps.interpolate().values)
print("Interpolated (method=quadratic):", s_gaps.interpolate(method='quadratic').round(2).values)
```

## 8. Data Types & Conversion

Getting the right data types is important for memory, performance, and correct operations.

```python
df_types = pd.DataFrame({
    'id': ['1', '2', '3', '4'],
    'price': ['29.99', '49.50', 'N/A', '19.99'],
    'quantity': [10, 20, 15, 25],
    'date_str': ['2024-01-15', '2024-02-20', '2024-03-10', '2024-04-05'],
    'is_active': ['True', 'False', 'True', 'True']
})
print("Original dtypes:")
print(df_types.dtypes)
print()
df_types
```

```python
# astype — cast to a different type
df_types['id'] = df_types['id'].astype(int)
df_types['is_active'] = df_types['is_active'].map({'True': True, 'False': False})

print("After astype conversions:")
print(df_types.dtypes)
```

```python
# pd.to_numeric — handles errors gracefully
# errors='coerce' turns unparseable values into NaN
df_types['price'] = pd.to_numeric(df_types['price'], errors='coerce')
print("Price after to_numeric (N/A → NaN):")
print(df_types[['price']])
print(f"dtype: {df_types['price'].dtype}")
```

```python
# pd.to_datetime — parse date strings
df_types['date'] = pd.to_datetime(df_types['date_str'])
print("Parsed dates:")
print(df_types[['date_str', 'date']])
print(f"\ndtype: {df_types['date'].dtype}")

# Extract components from datetime
print(f"\nYear: {df_types['date'].dt.year.values}")
print(f"Month: {df_types['date'].dt.month.values}")
print(f"Day name: {df_types['date'].dt.day_name().values}")
```

## 9. Basic String Operations

The `.str` accessor provides vectorized string methods on Series. Much faster than applying Python string methods row by row.

```python
names = pd.Series(['  Alice Smith  ', 'BOB JONES', 'charlie brown', 'Diana Prince', 'eve adams'])

print("Original:       ", names.values)
print("lower():        ", names.str.lower().values)
print("upper():        ", names.str.upper().values)
print("title():        ", names.str.title().values)
print("strip():        ", names.str.strip().values)
print("len():          ", names.str.len().values)
```

```python
# contains, startswith, endswith — returns boolean Series
names_clean = names.str.strip().str.title()
print("Contains 'a' (case-insensitive):")
print(names_clean.str.contains('a', case=False).values)

print("\nStarts with 'A':")
print(names_clean.str.startswith('A').values)
```

```python
# split — split strings into parts
print("Split by space:")
print(names_clean.str.split(' ', expand=True))

# Get first name only
print("\nFirst names:")
print(names_clean.str.split(' ').str[0].values)
```

```python
# replace, slice
emails = pd.Series(['alice@gmail.com', 'bob@yahoo.com', 'charlie@gmail.com'])
print("Replace gmail with outlook:")
print(emails.str.replace('gmail', 'outlook').values)

# Extract domain
print("\nDomains:")
print(emails.str.split('@').str[1].values)

# Slice first 5 characters
print("\nFirst 5 chars:")
print(emails.str[:5].values)
```

## 10. Basic GroupBy & Aggregation

`groupby` splits data into groups based on column values, applies a function to each group, and combines the results. This is the **split-apply-combine** pattern.

```python
np.random.seed(42)
df_gb = pd.DataFrame({
    'department': np.random.choice(['Sales', 'Engineering', 'HR', 'Marketing'], 20),
    'gender': np.random.choice(['M', 'F'], 20),
    'salary': np.random.randint(50000, 120000, 20),
    'experience': np.random.randint(1, 20, 20),
    'rating': np.random.uniform(3.0, 5.0, 20).round(1)
})
df_gb.head(10)
```

```python
# Basic groupby — single aggregation
print("Average salary by department:")
print(df_gb.groupby('department')['salary'].mean().round(0))
```

```python
# Multiple aggregations on one column
print("Salary stats by department:")
print(df_gb.groupby('department')['salary'].agg(['mean', 'min', 'max', 'count']))
```

```python
# Different aggregations per column
print("Mixed aggregations:")
result = df_gb.groupby('department').agg({
    'salary': 'mean',
    'experience': 'median',
    'rating': ['mean', 'max']
})
result
```

```python
# GroupBy with multiple keys
print("Average salary by department AND gender:")
df_gb.groupby(['department', 'gender'])['salary'].mean().round(0)
```

```python
# size() vs count() — size includes NaN, count doesn't
print("Group sizes:")
print(df_gb.groupby('department').size())
```

## 11. Merging, Joining & Concatenation

Combining DataFrames is one of the most common operations. pandas provides three main approaches:
- **`merge()`** — SQL-style joins on columns
- **`concat()`** — stack DataFrames vertically or horizontally
- **`join()`** — join on index

```python
# Setup: two related DataFrames
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'dept_id': [101, 102, 101, 103, 102]
})

departments = pd.DataFrame({
    'dept_id': [101, 102, 103, 104],
    'dept_name': ['Sales', 'Engineering', 'HR', 'Marketing']
})

print("Employees:")
print(employees)
print("\nDepartments:")
print(departments)
```

```python
# merge — inner join (only matching rows)
print("INNER join (default):")
print(employees.merge(departments, on='dept_id'))
```

```python
# merge — left, right, outer joins
print("LEFT join (all employees, matching depts):")
print(employees.merge(departments, on='dept_id', how='left'))

print("\nRIGHT join (all depts, matching employees):")
print(employees.merge(departments, on='dept_id', how='right'))

print("\nOUTER join (all from both):")
print(employees.merge(departments, on='dept_id', how='outer'))
```

```python
# merge on differently named columns
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'salary': [70000, 85000, 92000, 78000, 65000]
})

merged = employees.merge(salaries, left_on='emp_id', right_on='employee_id')
print("Merge on different column names:")
merged
```

## 12. Apply, Map & Replace

These methods let you transform values element-wise or row/column-wise.

```python
df_apply = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'score': [85, 62, 91],
    'grade_code': ['A', 'C', 'A']
})
df_apply
```

```python
# map — transform each element using a function or dict (Series only)
grade_map = {'A': 'Excellent', 'B': 'Good', 'C': 'Average', 'D': 'Poor'}
df_apply['grade_label'] = df_apply['grade_code'].map(grade_map)
print("After map:")
print(df_apply)
```

```python
# map with a function
df_apply['name_length'] = df_apply['name'].map(len)
df_apply['pass_fail'] = df_apply['score'].map(lambda x: 'Pass' if x >= 70 else 'Fail')
print("After map with functions:")
df_apply
```

```python
# apply — more flexible, works on Series or DataFrame
# On a Series: applies function to each element
print("apply on Series (double the score):")
print(df_apply['score'].apply(lambda x: x * 2).values)

# On a DataFrame with axis=1: applies function to each row
df_apply['summary'] = df_apply.apply(
    lambda row: f"{row['name']} scored {row['score']}", axis=1
)
print("\napply on rows:")
print(df_apply[['name', 'score', 'summary']])
```

```python
# replace — substitute specific values
s = pd.Series(['Yes', 'No', 'Yes', 'Maybe', 'No'])
print("Original:", s.values)
print("Replaced:", s.replace({'Yes': 1, 'No': 0, 'Maybe': -1}).values)
```

## Final Practice Challenge

Combine everything you've learned! Using this employee dataset:

1. Create a DataFrame with columns: `name`, `department`, `salary`, `hire_date`, `rating` (at least 10 rows, include some NaN values).
2. Inspect it: shape, dtypes, null counts, describe.
3. Clean it: fill missing ratings with the department mean, drop rows with missing names.
4. Add columns: `years_employed` (from `hire_date`), `salary_level` ('High'/'Low' based on median).
5. Group by department: compute average salary, max rating, and count.
6. Sort by salary descending and show the top 5.
7. Create a bar plot of average salary by department.

```python
# Solution: Final Challenge

# 1. Create the DataFrame
np.random.seed(42)
df_final = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank',
             'Grace', 'Henry', None, 'Ivy', 'Jack', 'Karen'],
    'department': ['Sales', 'Engineering', 'Sales', 'Engineering', 'HR', 'HR',
                   'Sales', 'Engineering', 'HR', 'Sales', 'Engineering', 'HR'],
    'salary': [70000, 95000, 82000, 88000, 65000, 72000,
               78000, 105000, 60000, 75000, 92000, 68000],
    'hire_date': pd.to_datetime([
        '2020-03-15', '2018-07-01', '2019-11-20', '2021-01-10', '2022-05-05',
        '2017-09-30', '2023-02-14', '2016-12-01', '2020-08-22', '2021-06-15',
        '2019-04-10', '2022-11-01'
    ]),
    'rating': [4.2, 4.8, 3.9, np.nan, 3.5, 4.0, np.nan, 4.9, 3.7, 4.1, np.nan, 3.8]
})

print("1. Created DataFrame:")
print(df_final)

# 2. Inspect
print(f"\n2. Shape: {df_final.shape}")
print(f"   Nulls per column:\n{df_final.isna().sum()}")
print(f"\n   Describe:\n{df_final.describe()}")

# 3. Clean
df_final = df_final.dropna(subset=['name'])
df_final['rating'] = df_final.groupby('department')['rating'].transform(
    lambda x: x.fillna(x.mean())
)
print(f"\n3. After cleaning — Nulls: {df_final.isna().sum().sum()}, Rows: {len(df_final)}")

# 4. Add columns
df_final['years_employed'] = (
    (pd.Timestamp('2024-06-01') - df_final['hire_date']).dt.days / 365.25
).round(1)
median_salary = df_final['salary'].median()
df_final['salary_level'] = np.where(df_final['salary'] >= median_salary, 'High', 'Low')
print(f"\n4. With new columns (median salary = ${median_salary:,.0f}):")
print(df_final[['name', 'salary', 'salary_level', 'years_employed']])

# 5. GroupBy
dept_summary = df_final.groupby('department').agg(
    avg_salary=('salary', 'mean'),
    max_rating=('rating', 'max'),
    count=('name', 'count')
)
print("\n5. Department summary:")
print(dept_summary)

# 6. Top 5 by salary
print("\n6. Top 5 earners:")
print(df_final.nlargest(5, 'salary')[['name', 'department', 'salary']])

# 7. Bar plot
dept_summary['avg_salary'].plot(
    kind='bar', title='Average Salary by Department',
    color='steelblue', edgecolor='black'
)
plt.ylabel('Average Salary ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
