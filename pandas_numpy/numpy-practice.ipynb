Masking : numpy.ma

```python
import numpy.ma as ma

arr = np.array([*range(15)])

print(arr)
print(ma.masked_equal(arr, value=10))           # will mask all values equal to 10 in the array
print(ma.masked_greater_equal(arr, value=10))   # will mask all values greater & equal to 10 in the array

print(ma.masked_inside(arr, 5, 11))             # masks values between a specified range (5 till 11)
print(ma.masked_outside(arr, 5, 11))            # masks values outside a specified range (5 till 11)


print(ma.masked_where(arr % 2 == 0, arr))       # mask all the even numbers in array

print("\n")
arr_inv = [1,2,3,4,np.inf,6,7,np.nan,9]
print(arr_inv)
print(ma.masked_invalid(arr_inv))               # masks invalid values like nan, infinity, etc
```

```python
## Vectorization example

array = np.array([*range(20)])
print(array)

def cube_if_even(num):
    """
    Returns cube of number if its is Even, else returns same number
    """
    if num%2 == 0:
        return num ** 3
    else:
        return num


vectorized_cube_if_even = np.vectorize(cube_if_even)

print(vectorized_cube_if_even(array))

    

```

```python
# defining custom data type

dt = np.dtype("i4")

arr1 = np.array(["1111","2222","3333"])
print(arr1, arr1.nbytes)

arr2 = np.array(["1111","2222","3333"], dtype=dt)      # specify the datatype as Int with length 4 bytes
print(arr2, arr2.nbytes)         # notice the change in datatype and total bytes                       
```
